### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


type_totals = df.groupby('Type 1')['Total'].mean().reset_index()


type_totals = type_totals.sort_values('Total', ascending=False)


sns.set(style="whitegrid")


plt.figure(figsize=(12, 6))
ax = sns.barplot(data=type_totals, x='Type 1', y='Total', palette='Set2')


plt.title('Average Total Stats by Primary Pokémon Type')
plt.xlabel('Primary Type')
plt.ylabel('Average Total Stat')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    data=df,
    x='Speed',
    y='Total',
    hue='Legendary',
    palette='Set1',
    alpha=0.7
)


plt.title('Pokémon: Total Stats vs. Speed')
plt.xlabel('Speed')
plt.ylabel('Total Stats')
plt.legend(title='Legendary')
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')

# Calculate Generation-wise average for Total stats
df["Generation"] = df["Generation"].astype(int)
gen_avg = df.groupby(["Generation"]).mean(numeric_only= True)



sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
sns.lineplot(data=gen_avg, x='Generation', y='Total', marker='o', color='mediumseagreen')

plt.title('Average Total Stats by Pokémon Generation')
plt.xlabel('Generation')
plt.ylabel('Average Total Stats')
plt.xticks(df['Generation'])
plt.tight_layout()
plt.show()

```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
import pandas as pd
df = pd.read_csv("W1_Pokemon.csv")
gens = df.groupby(["Generation" , "Legendary"])
df2 = gens.get_group((4, False))
filt = ((df["Type 1"] == "Ghost") | (df["Type 2"] == "Ghost"))
df3 = df2.loc[filt].sort_values(by= "Speed" , ascending= False).Name
df3.iloc[1]
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
import pandas as pd
df = pd.read_csv("W1_Pokemon.csv")
Charizard_def = df.loc[(df["Name"] == "Charizard") , "Defense"].values[0]
Charizard_atk = df.loc[(df["Name"] == "Charizard") , "Attack"].values[0]
filt = (((df["Defense"] >= Charizard_def) & (df["Attack"] <= Charizard_atk)) & (df["Legendary"] == False))
df.loc[filt , "Name"].count()

```

### How many generation-3, non-legendary pokemons have higher HP than the weakest generation-6, legendary pokemon?

```python
gen3 = df.groupby(["Generation"]).get_group(3)
gen6 = df.groupby(["Generation"]).get_group(6)
filt = gen6["Legendary"] == True 
hp = gen6.loc[filt , "HP"].min()
filt2 = (gen3["HP"] > hp) & (gen3["Legendary"] == False)
gen3.loc[filt2, "Name"].count()
```
