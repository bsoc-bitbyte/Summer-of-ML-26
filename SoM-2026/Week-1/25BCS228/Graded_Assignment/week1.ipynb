### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
#your code here
secGhost = pok[(pok['Generation'] == 4) & (pok['Legendary']==False) & ((pok['Type 1']=='Ghost') | (pok['Type 2']=='Ghost'))]
an7 = secGhost.sort_values(by='Speed', ascending=False).index[1]
an7
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
#your code here
char_atk = pok.loc['Charizard', 'Attack']
char_def = pok.loc['Charizard', 'Defense']
an8 = pok[(pok['Legendary']==False) & (pok.Defense > char_def) & (pok.Attack < char_atk)]
len(an8)
```
