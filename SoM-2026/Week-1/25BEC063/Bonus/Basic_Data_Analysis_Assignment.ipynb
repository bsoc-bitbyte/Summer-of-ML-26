### 16) Which profession has youngest people(apart from student)? Show the trend of mean age of people with profession as the seperating factor.

```python
import seaborn as sns
import matplotlib.pyplot as plt

mean_age = df.groupby('occupation')['age'].mean()

sns.barplot(
    x=mean_age.values, 
    y=mean_age.index,
)

plt.title("Trend of Mean Age v/s Profession")
plt.xlabel("Mean Age")
plt.ylabel("Profession")
plt.tight_layout()
plt.show()
```
