### 1. How many pokemons have 'Mega' in their name?

```python
mega_count = df[df.index.str.contains('Mega')].shape[0]
print(f"Mega pokemons: {mega_count}")
```

### 2. What is the standard deviation of Sp. Def. in the dataset ?

```python

print(f"Sp. Def std: {df['Sp. Def'].std()}")
```

### 3. What percentage (upto 3 decimal places) of pokemons are legendary ?

```python

legendary_pct = (df['Legendary'].sum() / len(df)) * 100
print(f"Legendary %: {legendary_pct:.3f}")
```

### 4. Name the pokemon(s) with Maximum Defense.

```python

max_def = df['Defense'].max()
print(df[df['Defense'] == max_def][['Defense']])
```

### 5. Which poison pokemon has the strongest attack ?

```python

poison = df[df['Type 1'] == 'Poison']
print(poison[poison['Attack'] == poison['Attack'].max()][['Attack']])
```

### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python

sns.histplot(df['Speed'], kde=True)
plt.title('Speed Distribution')
plt.show()
```

```python

sns.scatterplot(data=df, x='Attack', y='Defense', hue='Legendary')
plt.title('Attack vs Defense')
plt.show()
```

```python

type_stats = df.groupby('Type 1')['Total'].mean().sort_values()
sns.barplot(x=type_stats.values, y=type_stats.index)
plt.title('Avg Total Stats by Type 1')
plt.tight_layout()
plt.show()
```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python

ghost_gen4 = df[
    (df['Type 1'] == 'Ghost') &
    (df['Generation'] == 4) &
    (df['Legendary'] == False)
].sort_values('Speed', ascending=False)
print(ghost_gen4.iloc[1][['Speed']])
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python

charizard = df.loc['Charizard']
result = df[
    (df['Defense'] > charizard['Defense']) &
    (df['Attack'] < charizard['Attack']) &
    (df['Legendary'] == False)
]
print(f"Count: {len(result)}")
```

### Which pokemon has the highest combined value of Attack and Sp. Atk ?

```python
df['Atk_Combined'] = df['Attack'] + df['Sp. Atk']
print(df[df['Atk_Combined'] == df['Atk_Combined'].max()][['Atk_Combined']])
```

### Which type of legendary pokemons are the most common?

```python

legendary_df = df[df['Legendary'] == True]
print(legendary_df['Type 1'].value_counts().idxmax())
```

### How many generation-3, non-legendary pokemons have higher HP than the weakest generation-6, legendary pokemon?

```python
weakest_gen6_leg_hp = df[
    (df['Generation'] == 6) & (df['Legendary'] == True)
]['HP'].min()

gen3_nonleg = df[
    (df['Generation'] == 3) &
    (df['Legendary'] == False) &
    (df['HP'] > weakest_gen6_leg_hp)
]
print(f"Count: {len(gen3_nonleg)}")
```

### Print out the third slowest pokemon(s) in the dataset.

```python

third_slowest_speed = df['Speed'].sort_values().unique()[2]
print(df[df['Speed'] == third_slowest_speed][['Speed']])
```

### Which pokemon type has the highest average HP?
