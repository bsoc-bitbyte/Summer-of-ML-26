### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python
df = pd.read_csv('W1_Pokemon.csv')


type_totals = df.groupby('Type 1')['Total'].mean().reset_index()


type_totals = type_totals.sort_values('Total', ascending=False)


sns.set(style="whitegrid")


plt.figure(figsize=(12, 6))
ax = sns.barplot(data=type_totals, x='Type 1', y='Total', palette='Set2')


plt.title('Average Total Stats by Primary Pokémon Type')
plt.xlabel('Primary Type')
plt.ylabel('Average Total Stat')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv('W1_Pokemon.csv')


sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    data=df,
    x='Speed',
    y='Total',
    hue='Legendary',
    palette='Set1',
    alpha=0.7
)


plt.title('Pokémon: Total Stats vs. Speed')
plt.xlabel('Speed')
plt.ylabel('Total Stats')
plt.legend(title='Legendary')
plt.tight_layout()
plt.show()
```

```python
df = pd.read_csv('W1_Pokemon.csv')
gen_avg = df.groupby("Generation")["Total"].mean().reset_index()


sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
sns.lineplot(data=gen_avg, x='Generation', y='Total', marker='o', color='mediumseagreen')

plt.title('Average Total Stats by Pokémon Generation')
plt.xlabel('Generation')
plt.ylabel('Average Total Stats')
plt.xticks(gen_avg['Generation'])
plt.tight_layout()
plt.show()
```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python

non_legen_ghost_fthgen = df[(df["Generation"]==4) & (df["Legendary"]==False) & ((df["Type 1"]=="Ghost") | (df["Type 2"]=="Ghost"))].sort_values(by="Speed",ascending=False)
print(non_legen_ghost_fthgen.iloc[1]["Name"])
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
defence_of_charizard = df[(df["Name"]=="Charizard")]["Defense"].iloc[0]
attack_of_charizard = df[(df["Name"]=="Charizard")]["Attack"].iloc[0]

sdbwa = len(df[(df["Legendary"]==False) & (df["Attack"]<attack_of_charizard) & (df["Defense"]>defence_of_charizard)])
print(sdbwa)
```

### Which pokemon has the highest combined value of Attack and Sp. Atk ?

### Which type of legendary pokemons are the most common?

```python
legendary = df[df["Legendary"]==True]

maxx=0
ans =""

for i in df["Type 1"].dropna().unique():
    cnt = len(legendary[legendary["Type 1"]==i]) + len(legendary[legendary["Type 2"]==i])

    if cnt>maxx:
        maxx=cnt
        ans = i
print(ans)
```

### How many generation-3, non-legendary pokemons have higher HP than the weakest generation-6, legendary pokemon?

```python

gen_and_nl = df[(df["Generation"]==3) & (df["Legendary"]==False)]
gens_and_l = df[(df["Generation"]==6) & (df["Legendary"]==True)]["HP"].min()

print(len(gen_and_nl[gen_and_nl["HP"] > gens_and_l]))
```
