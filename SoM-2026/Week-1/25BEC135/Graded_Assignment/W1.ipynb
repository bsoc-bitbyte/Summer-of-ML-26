### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
#your code here
ghost_pokemon=df[
    ((df['Type 1'] == 'Ghost') | (df['Type 2'] == 'Ghost')) &
    (df['Generation'] == 4) &
    (df['Legendary'] == False)
]
ghost_pokemon.sort_values('Speed', ascending=False).iloc[1]['Name']
```
