```python
east_asian_countries = df1[df1['Regional indicator'] == 'East Asia']

top5 = east_asian_countries.sort_values(
    'Generosity',
    ascending=False
)

print(top5[['Country name', 'Generosity']].iloc[0:5])
```

```python
plt.plot(
    top5['Country name'],
    top5['Ladder score'],
    marker='o',
    color='red',

    linewidth=2,
    markersize=10,
    markerfacecolor='blue',
    markeredgecolor='green',
)

plt.title('Ladder Score of Top 5 East Asian Countries by Generosity')
plt.xlabel('Country')
plt.ylabel('Ladder Score')

plt.grid(True)
plt.show()
```

```python
countries = [
    'China',
    'Nepal',
    'Bangladesh',
    'Pakistan',
    'Myanmar',
    'India',
    'Afghanistan'
]

df_2021 = df1[df1['Country name'].isin(countries)]

df_2021
```

```python
plt.figure()

plt.scatter(
    df_2021['Country name'],
    df_2021['Logged GDP per capita']
)

plt.title('Country vs Logged GDP per Capita')
plt.xlabel('Country')
plt.ylabel('Logged GDP per Capita')

plt.xticks(rotation=45,ha='right')

plt.show()
```
