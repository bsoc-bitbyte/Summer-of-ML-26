### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


type_totals = df.groupby('Type 1')['Total'].mean().reset_index()


type_totals = type_totals.sort_values('Total', ascending=False)


sns.set(style="whitegrid")


plt.figure(figsize=(12, 6))
ax = sns.barplot(data=type_totals, x='Type 1', y='Total', palette='Set2')


plt.title('Average Total Stats by Primary Pokémon Type')
plt.xlabel('Primary Type')
plt.ylabel('Average Total Stat')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    data=df,
    x='Speed',
    y='Total',
    hue='Legendary',
    palette='Set1',
    alpha=0.7
)


plt.title('Pokémon: Total Stats vs. Speed')
plt.xlabel('Speed')
plt.ylabel('Total Stats')
plt.legend(title='Legendary')
plt.tight_layout()
plt.show()

```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')

# Calculate Generation-wise average for Total stats
gen_avg = df.groupby('Generation')['Total'].mean().reset_index()


sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
sns.lineplot(data=gen_avg, x='Generation', y='Total', marker='o', color='mediumseagreen')

plt.title('Average Total Stats by Pokémon Generation')
plt.xlabel('Generation')
plt.ylabel('Average Total Stats')
plt.xticks(gen_avg['Generation'])
plt.tight_layout()
plt.show()

```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
import pandas as pd
df = pd.read_csv("W1_Pokemon.csv")

is_gen_4 = df['Generation'] == 4
is_nonlegend=df['Legendary']==False
is_ghost=(df['Type 1']=='Ghost') | (df['Type 2']=='Ghost')

is_filtered=df[is_gen_4&is_ghost & is_nonlegend]
sorted_def=is_filtered.sort_values(by='Speed', ascending=False)

second_fastest = sorted_def.iloc[1]['Name']
print(second_fastest)
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
charizard_attack = df[df['Name'] == 'Charizard']['Attack'].values[0]
charizard_defense = df[df['Name'] == 'Charizard']['Defence'].values[0]

is_nonlegendary = df['Legendary'] == False
stronger_defense = df['Defense'] > charizard_defense
weaker_attack = df['Attack'] < charizard_attack

filtered_pokemon = df[is_nonlegendary & stronger_defense & weaker_attack]

final_count = len(filtered_pokemon)

print(f"Number of non-legendary Pokemon with stronger defense but weaker attack than Charizard: {final_count}")
```
