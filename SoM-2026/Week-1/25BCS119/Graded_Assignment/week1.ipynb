### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python
#your code here
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


type_totals = df.groupby('Type 1')['Total'].mean().reset_index()


type_totals = type_totals.sort_values('Total', ascending=False)


sns.set(style="whitegrid")


plt.figure(figsize=(12, 6))
ax = sns.barplot(data=type_totals, x='Type 1', y='Total', palette='Set2')


plt.title('Average Total Stats by Primary Pokémon Type')
plt.xlabel('Primary Type')
plt.ylabel('Average Total Stat')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


fig,(ax2,ax1)=plt.subplots(1,2,figsize=(10,10))
sns.histplot(data=df,x='Attack',element='step',hue='Type 1',ax=ax1)
sns.histplot(data=df,x='Speed',element='step',hue='Type 2',ax=ax2)
```

```python
#your code here
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')


sns.set(style="whitegrid")


plt.figure(figsize=(10, 6))
ax = sns.scatterplot(
    data=df,
    x='Speed',
    y='Total',
    hue='Legendary',
    palette='Set1',
    alpha=0.7
)

sns.pairplot(df,hue='Type 1')
plt.title('Pokémon: Total Stats vs. Speed')
plt.xlabel('Speed')
plt.ylabel('Total Stats')
plt.legend(title='Legendary')
plt.tight_layout()
plt.show()

```

```python
#your code here
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('W1_Pokemon.csv')



gen_avg=df.groupby('Generation')['Total'].mean().reset_index()

sns.set(style="darkgrid")


plt.figure(figsize=(10, 6))
sns.lineplot(data=gen_avg, x='Generation', y='Total', marker='o', color='mediumseagreen',label='Average Total Stats')

plt.title('Average Total Stats by Pokémon Generation')
plt.xlabel('Generation')
plt.ylabel('Average Total Stats')
plt.xticks(gen_avg['Generation'])
plt.tight_layout()
plt.legend()
plt.show()

```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
lst=[]
for name in df.index:
    if ((df.loc[name,'Type 1']=='Ghost' or df.loc[name,'Type 2']=='Ghost') and df.loc[name,'Generation'] == 4 and
        df.loc[name,'Legendary']==False):
        lst.append(((df.loc[name,'Speed'],name)))
lst.sort(reverse=True)
print(lst[1][1])
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
df.drop('#',axis=1,inplace=True)
df.set_index('Name',inplace=True)
c=0
catk = df.loc['Charizard', 'Attack']
cdef= df.loc[ 'Charizard', 'Defense']
for name in df.index:
    if df.loc[name,'Attack']<catk and df.loc[name,'Defense']<cdef and df.loc[name,'Legendary']==False:
        c=c+1
print(c)
```
