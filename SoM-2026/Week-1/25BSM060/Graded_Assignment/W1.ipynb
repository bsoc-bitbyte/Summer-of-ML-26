### 6. Using seaborn make different types of plots, observe the trend and answer the questions given in the form.

```python
plt.figure(figsize=(14, 5))
sns.countplot(data=df, x='Type 1', order=df['Type 1'].value_counts().index)
plt.title('Count of Pokemons by Type 1')
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Legendary', y='Total')
plt.title('Total Stats: Legendary vs Non-Legendary')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x='Attack', y='Defense', hue='Legendary')
plt.title('Attack vs Defense')
plt.show()
```

### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
ghost_gen4 = df[
    (df['Type 1'] == 'Ghost') & 
    (df['Generation'] == 4) & 
    (df['Legendary'] == False)
]
ghost_sorted = ghost_gen4.sort_values('Speed', ascending=False)
print(ghost_sorted[['Speed']])
print("\n2nd fastest:", ghost_sorted.index[1])
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?

```python
charizard = df.loc['Charizard']
print("Charizard -> Defense:", charizard['Defense'], "| Attack:", charizard['Attack'])

result = df[
    (df['Legendary'] == False) & 
    (df['Defense'] > charizard['Defense']) & 
    (df['Attack'] < charizard['Attack'])
]
print("Count:", len(result))
```

### Which type of legendary pokemons are the most common?

```python
leg_df = df[df['Legendary'] == True]
print(leg_df['Type 1'].value_counts())

leg_df = df[df['Legendary'] == True]
print(leg_df['Type 1'].value_counts().idxmax())
```

### How many generation-3, non-legendary pokemons have higher HP than the weakest generation-6, legendary pokemon?

```python
gen6_leg = df[(df['Generation'] == 6) & (df['Legendary'] == True)]
weakest_hp = gen6_leg['HP'].min()
print("Weakest Gen6 legendary HP:", weakest_hp)

gen3_nonleg = df[(df['Generation'] == 3) & (df['Legendary'] == False)]
count = len(gen3_nonleg[gen3_nonleg['HP'] > weakest_hp])
print("Count:", count)
```

### Print out the third slowest pokemon(s) in the dataset.

```python
sorted_speed = df.sort_values('Speed')
third_slowest_speed = sorted_speed['Speed'].unique()[2]
third_slowest = sorted_speed[sorted_speed['Speed'] == third_slowest_speed]
print("Speed value:", third_slowest_speed)
print("Pokemon(s):", third_slowest.index.tolist())
```

### Which pokemon type has the highest average HP?

```python
avg_hp_by_type = df.groupby('Type 1')['HP'].mean().sort_values(ascending=False)
print(avg_hp_by_type)
print("\nType with highest avg HP:", avg_hp_by_type.idxmax())
```
