### 14) What is the age with least occurrence?

```python
age_counts = users['age'].value_counts()
least_occ = age_counts[age_counts == age_counts.min()]
print("Age(s) with least occurrence:")
print(least_occ)
```

### 16) Which profession has youngest people(apart from student)? Show the trend of mean age of people with profession as the seperating factor.

```python
mean_age_by_occ = users.groupby('occupation')['age'].mean().sort_values()
print("Mean age by occupation (ascending):")
print(mean_age_by_occ)

# Excluding 'student'
youngest_profession = mean_age_by_occ.drop('student').idxmin()
print("\nYoungest profession (excl. student):", youngest_profession)
print("Mean age:", round(mean_age_by_occ[youngest_profession], 2))
```
