```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df1 = pd.read_csv('world-happiness-report-2021.csv')
print('Dataset loaded successfully!')
print('Shape:', df1.shape)
```

```python
print('Missing values per column:')
print(df1.isnull().sum())
```

```python
western_europe = df1[df1['Regional indicator'] == 'Western Europe']
print('Number of unique countries in Western Europe:', len(western_europe))
print('\nCountries:')
print(western_europe['Country name'].tolist())
western_europe[['Country name', 'Ladder score']]
```

```python
india = df1[df1['Country name'] == 'India']
print('India Happiness (Ladder) Score in 2021:', india['Ladder score'].values[0])
india[['Country name', 'Regional indicator', 'Ladder score']]
```

```python
east_asia = df1[df1['Regional indicator'] == 'East Asia']
top5_east_asia = east_asia.sort_values('Generosity', ascending=False).head(5)
print('Top 5 East Asian Countries by Generosity:')
top5_east_asia[['Country name', 'Generosity', 'Ladder score']]
```

```python
plt.figure(figsize=(10, 5))
plt.plot(top5_east_asia['Country name'], top5_east_asia['Ladder score'],
         marker='o', color='steelblue', linewidth=2, markersize=8)
plt.title('Ladder Score of Top 5 East Asian Countries (by Generosity)', fontsize=13)
plt.xlabel('Country')
plt.ylabel('Ladder Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

```python
countries = ['China', 'Nepal', 'Bangladesh', 'Pakistan', 'Myanmar', 'India', 'Afghanistan']
df_2021 = df1[df1['Country name'].isin(countries)].copy()
print('df_2021:')
df_2021[['Country name', 'Logged GDP per capita', 'Ladder score']]
```

```python
plt.figure(figsize=(10, 6))

colors = ['red', 'blue', 'green', 'orange', 'purple', 'brown', 'pink']
for i, (_, row) in enumerate(df_2021.iterrows()):
    plt.scatter(row['Logged GDP per capita'], row['Ladder score'],
                color=colors[i], s=150, label=row['Country name'])
    plt.annotate(row['Country name'],
                 (row['Logged GDP per capita'], row['Ladder score']),
                 textcoords='offset points', xytext=(6, 5), fontsize=9)

plt.title('Countries vs Logged GDP per Capita (2021)', fontsize=13)
plt.xlabel('Logged GDP per Capita')
plt.ylabel('Ladder Score (Happiness)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
