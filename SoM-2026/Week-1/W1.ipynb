### 7. Which is the second fastest non-legendary 'Ghost' type pokemon from 4th generation ?

```python
#your code here
import pandas as pd
f=pd.read_csv("W1_Pokemon.csv")
ghost=f[((f['Type 1']=='Ghost')|(f['Type 2']=='Ghost'))
&(f['Generation']==4)
&(f['Legendary']==False)]
ghost=ghost.sort_values('Speed',ascending=False)
print(ghost.iloc[1]['Name'])
```

### 8. How many non-legendary pokemons have stronger defence but weaker attack than Charizard?
